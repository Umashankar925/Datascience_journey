What is datatype?

A data type in Python defines what kind of value a variable holds and what operations can be performed on that value. Python has several built-in data types such as integers, floating-point numbers, strings, booleans, lists, tuples, sets, and dictionaries.

Why are datatypes are important?

Data types help Python determine how a value should be stored and what operations can be performed on it.

Basic datatypes are:
    1.integer
    2.float
    3.string
    4.bollean
    

```python
# 1. integer
age = 23
print(f"age = {age}")
print(type(age))
```

```python
# 2.float point values
weight = 83.54
print(f"weight = {weight}")
print(type(weight))
```

```python
# 3. string
course = 'Datascience'
print(f"course = {course}")
print(type(course))
```

```python
# 4. boolean values
is_student = True
print(type(is_student))
```

Problem 13: Monthly Salary

Create:

HRA = 20%

PF = 5%

Gross Salary = Basic + HRA + DA

Net Salary = Gross Salary - PF

Print all values.

```python
basic_salary = 35000
HRA = basic_salary *(20/100)
DA = basic_salary *(10/100)
PF = basic_salary *(5/100)
Gross_salary = basic_salary+HRA+DA
Net_salary = Gross_salary - PF
print(f"Basic salary is {basic_salary}")
print(f"HRA is {HRA}")
print(f"DA is {DA}")
print(f"PF is {PF}")
print(f"Gross salary is {Gross_salary}")
print(f"Net salary is {Net_salary}")
```
