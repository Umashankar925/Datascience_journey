Pandas is a powerful data manipulation library in python, widely used for data analysis and 
data cleaning .
A series is a one dimensional array like object,while dataframe is a two dimensional array like object which contains multiple rows and columns.

```python
import pandas as pd
```

```python
#series
data = [1,2,3,4,5]
series = pd.Series(data)
print(series)

```

```python
#creating a series from dictionary.
data = {'a':1,'b':2,'c':3,'d':4,'e':5}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data,index = index)
```

```python
#DataFrame
#contains multiple rows and columns.


#creating a dataframe from a dictionary of list.
data = {
    'name':['krish','ajay','varun'],'age':[10,20,30],'city':['bangalore','pune','chennai']
}
df = pd.DataFrame(data)
df
```

```python
#creating a dataframe from a list of dictionary.
data = [
    {'name':'krish','age':30,'city':'bangalore'},
    {'name':'ajay','age':40,'city':'pune'},
    {'name':'varun','age':50,'city':'chennai'}
]
df = pd.DataFrame(data)
df
```

```python
data
```

```python
#accessing data from dataframe.
df['name']
```

```python
df.loc[0] #shows the details of the row which is at zeroth index.
```

```python
df.iloc[1,2] #takes the rows and column indexes.
```

```python
#accessing a specified element using.
df.at[2,'age']
```

```python
df.at[1,'city']
```

```python
df.at[0,'name']
```

```python
#accessing a specified element using iat.
df.iat[2,2] #here we will be giving both row index and column index.
```
