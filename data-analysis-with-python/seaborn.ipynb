seaborn helps in creating complex visualizations with just a few lines of code.

```python
import seaborn as sns
```

```python
#Basic plotting with seaborn

tips = sns.load_dataset('tips')
tips
```

```python
#creating a scatter plot.
import matplotlib.pyplot as plt
df = tips
sns.scatterplot(data = df,x = 'total_bill',y = 'tip',color = 'purple')
plt.title('scatter plot of total bill vs tips')
plt.show()
```

```python
#line plot.
sns.lineplot(data = df,x = "total_bill",y = "tip",color = 'purple',marker = '.',markersize = 10)
plt.show()
```

```python
#categorical plots
#1.box plot -- it is used to identify the outliers.
sns.boxplot(data = df,x = "day",y  = "total_bill",color = 'purple')
plt.show()
```

```python
#2.bar plot.
sns.barplot(data = df,x = 'day',y = 'tip',color = 'purple')
plt.show()
```

```python
#3.violin plot
#it shows how the data is distributed for different categories.
sns.violinplot(data = df,x = 'day',y='total_bill')
```

```python
#4.histogram
sns.histplot(data = df['total_bill'],bins = 10,kde = True)
#kde - kerenel density estimator forms the curve for the given data.
```

```python
#5.pair plot
#displays the relationship between every pair of numerical value in a dataset in a single figure.
sns.pairplot(data =df)
plt.show()
```

```python
#6.heatmap
#It is a graphical representation of data using colors.
corr = df[['total_bill','tip','size']].corr()
corr
sns.heatmap(corr,annot=True,cmap = 'viridis')
```
