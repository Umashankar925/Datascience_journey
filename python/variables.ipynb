Problem 15: Simple Interest

Take:

Principal
Rate
Time

Formula:

SI = (P × R × T) / 100

```python
P =int(input("Enter the principal amount:"))#principal means the original amount borrowed.
R = float(input("Enter the interest rate:"))#5% is the interest rate.
T = int(input("Enter the tenure:"))#time is nothing for how long you will be taking this amount.

SI = (P*R*T)/100
total_amount = SI+P
print(f"simple interest = {SI}")
print(f"total amount after the interest will be {total_amount}")

```

Problem 17: BMI Calculator

Take:

Weight (kg)
Height (meters)

Formula:

BMI = weight / (height × height)

```python
weight = float(input("Enter the weight:"))
Height = float(input("Enter the height in centimeters:"))
BMI = (weight/(Height*Height))*10000
print(f"BMI is {BMI}")
if BMI<18.5:
    print(f"Under weight")
elif BMI>=18.5 and BMI<= 24.9:
    print(f"Healthy weight")
elif BMI>=25.0 and BMI<=29.9:
    print(f"Over weight")
else:
    print(f"obesity")
```
